```python
import os
from pathlib import Path

print("Running Environment & Path Integrity Test...")

try:
    BASE_DIR = Path.cwd().parent()
    DATA_FILE = BASE_DIR / "data" / "consumer_basket_2020_2023.csv"
    FIGURES_DIR = BASE_DIR / "figures"

    print(f"  * Working Directory : {BASE_DIR}")
    print(f"  * Target Data File  : {DATA_FILE}")

    if not DATA_FILE.exists():
        raise FileNotFoundError(
            f"Dataset not found at '{DATA_FILE}'. "
            f"Ensure the 'data' folder is in the same directory as this notebook."
        )

    print("  [PASS] File paths and environment resolved successfully.\n")
except Exception as e:
    print(f"  [FAIL] Environment Error: {e}\n")
```

```python
import pandas as pd

print("Running CSV Schema & Data Validation Test...")

NECESSITIES = ["food_at_home", "housing", "transportation", "healthcare"]
LUXURIES = [
    "food_away_from_home",
    "alcoholic_beverages",
    "apparel_and_services",
    "entertainment",
    "personal_care",
    "reading",
]
REQUIRED_BASE_COLS = ["year", "inflation_pct", "total_expenditure"]

try:
    df_raw = pd.read_csv(DATA_FILE)
    print(
        f"  * Raw Data Loaded: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns"
    )

    # Check for mandatory base columns
    missing_base = [
        col for col in REQUIRED_BASE_COLS if col not in df_raw.columns
    ]
    if missing_base:
        raise KeyError(f"Missing mandatory base columns: {missing_base}")

    # Check for basket columns
    missing_nec = [col for col in NECESSITIES if col not in df_raw.columns]
    missing_lux = [col for col in LUXURIES if col not in df_raw.columns]

    if missing_nec or missing_lux:
        raise KeyError(
            f"Missing basket categories in CSV -> Necessities missing: {missing_nec}, Luxuries missing: {missing_lux}"
        )

    # Verify data types are numeric
    all_target_cols = REQUIRED_BASE_COLS + NECESSITIES + LUXURIES
    non_numeric = [
        col
        for col in all_target_cols
        if not pd.api.types.is_numeric_dtype(df_raw[col])
    ]
    if non_numeric:
        raise TypeError(f"Columns contain non-numeric data: {non_numeric}")

    print(
        "  [PASS] All expected columns are present with valid numeric types.\n"
    )
except Exception as e:
    print(f"  [FAIL] CSV Validation Error: {e}\n")
```
